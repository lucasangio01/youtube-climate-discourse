# Data preprocessing workflow

1. __Split text__ --> for each video, split text into chunks of about 250 words, each chunk will end up on one different row (for model limits).
2. __Add punctuation__ --> add punctuation to each chunk with a punctuation restoration model (transformer)
3. __Check similarity to topic__ --> for each chunk (row) compute semantic similarity with some phrases (like "climate", "global warming", etc)
3. __Filter by similarity__ -->only keep the chunks that have semantic similarity > threshold (like 0.3 or 0.4 for cosine similarity)
4. __Tokenize__ -->tokenize filtered chunks for stance detection
